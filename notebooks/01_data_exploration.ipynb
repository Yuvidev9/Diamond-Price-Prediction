**Step 1 — Import Pandas and load the dataset**

```python
# Step 1: Import the required library
import pandas as pd

# Step 2: Load the dataset
df = pd.read_csv("../data/diamonds.csv")

# Step 3: Display the first 5 rows
df.head()
```

**Step 2 — Check the dataset structure**

```python
# Check the number of rows and columns
print("Dataset shape:", df.shape)

# Display column names
print("\nColumn names:")
print(df.columns.tolist())

# Check data types and non-null values
print("\nDataset information:")
df.info()
```

**Step 3 — Check missing values and duplicates**

```python
# Check missing values in each column
print("Missing values:")
print(df.isnull().sum())

# Check duplicate rows
print("\nNumber of duplicate rows:", df.duplicated().sum())
```

**Step 4 — Check the statistical summary**

```python
# Summary of numerical columns
df.describe()
```

**Step 5 — Understand the dataset using statistical analysis**

```python
# Analyze diamond prices

print("Minimum price:", df["price"].min())
print("Maximum price:", df["price"].max())
print("Average price:", df["price"].mean())
print("Median price:", df["price"].median())
```

```python
# Visualize the distribution of diamond prices

import matplotlib.pyplot as plt
import seaborn as sns

# Plot the distribution of diamond prices

plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="price", bins=50, kde=True)

plt.title("Distribution of Diamond Prices")
plt.xlabel("Price")
plt.ylabel("Number of Diamonds")

plt.show()
```

**Step 6 — Remove the unnecessary column**

```python
# Drop the unnecessary index column

df = df.drop(columns=["Unnamed: 0"], errors="ignore")

# Verify the updated dataset
print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())
```
